```python
# ============================================================
# ✈️ AI TRAVEL PLANNER USING TRANSFORMER
# Google Colab + Gradio
# Transformer Model: FLAN-T5
# ============================================================

# STEP 1: Install libraries
!pip -q install transformers sentencepiece accelerate gradio

# ============================================================
# STEP 2: Import libraries
# ============================================================

import torch
import gradio as gr
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# ============================================================
# STEP 3: Load Transformer Model
# ============================================================

print("Loading Transformer model...")

model_name = "google/flan-t5-small"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

print("✅ Transformer model loaded!")


# ============================================================
# STEP 4: AI TRAVEL PLANNER FUNCTION
# ============================================================

def generate_travel_plan(
    destination,
    days,
    budget,
    travelers,
    interests,
    food,
    travel_style
):

    # Check empty fields
    if not destination:
        return "❌ Please enter a destination."

    if not days:
        return "❌ Please enter the number of days."

    if not budget:
        return "❌ Please enter your budget."

    if not travelers:
        return "❌ Please enter the number of travelers."

    if not interests:
        interests = "general sightseeing"

    if not food:
        food = "local food"

    if not travel_style:
        travel_style = "comfortable"


    # ========================================================
    # TRANSFORMER PROMPT
    # ========================================================

    prompt = f"""
You are an expert AI Travel Planner.

Create a personalized travel itinerary.

Destination: {destination}
Number of days: {days}
Budget: {budget}
Number of travelers: {travelers}
Interests: {interests}
Food preference: {food}
Travel style: {travel_style}

Create a day-by-day itinerary.

For every day provide:

Morning:
Afternoon:
Evening:
Food:
Transportation:
Estimated expense:

At the end provide:

1. Best places to visit
2. Travel tips
3. Packing checklist
4. Approximate total expense

Keep the answer simple, practical and well organized.
"""

    # ========================================================
    # TOKENIZATION
    # ========================================================

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=512
    )

    # ========================================================
    # GENERATE USING TRANSFORMER
    # ========================================================

    with torch.no_grad():

        output = model.generate(
            **inputs,
            max_new_tokens=700,
            do_sample=True,
            temperature=0.7,
            top_p=0.9
        )

    # ========================================================
    # CONVERT OUTPUT TO TEXT
    # ========================================================

    result = tokenizer.decode(
        output[0],
        skip_special_tokens=True
    )

    return result


# ============================================================
# STEP 5: CREATE GRADIO INTERFACE
# ============================================================

with gr.Blocks(
    title="AI Travel Planner"
) as app:

    # --------------------------------------------------------
    # TITLE
    # --------------------------------------------------------

    gr.Markdown(
        """
        # ✈️ AI Travel Planner

        ### Personalized travel itinerary using a Transformer model

        Enter your travel preferences below and let AI create
        a personalized day-by-day travel plan.
        """
    )

    # --------------------------------------------------------
    # INPUT SECTION
    # --------------------------------------------------------

    with gr.Row():

        destination = gr.Textbox(
            label="📍 Destination",
            placeholder="Example: Dubai"
        )

        days = gr.Number(
            label="📅 Number of Days",
            value=3
        )

    with gr.Row():

        budget = gr.Textbox(
            label="💰 Budget",
            placeholder="Example: ₹50,000"
        )

        travelers = gr.Number(
            label="👥 Number of Travelers",
            value=2
        )

    interests = gr.Textbox(
        label="❤️ Interests",
        placeholder="Example: beaches, shopping, adventure"
    )

    food = gr.Textbox(
        label="🍽️ Food Preference",
        placeholder="Example: vegetarian, halal, local food"
    )

    travel_style = gr.Dropdown(
        choices=[
            "Budget",
            "Comfortable",
            "Luxury",
            "Adventure",
            "Relaxed"
        ],
        value="Comfortable",
        label="✨ Travel Style"
    )

    # --------------------------------------------------------
    # BUTTON
    # --------------------------------------------------------

    generate_button = gr.Button(
        "✈️ Generate Travel Plan",
        variant="primary"
    )

    # --------------------------------------------------------
    # OUTPUT
    # --------------------------------------------------------

    output = gr.Markdown(
        label="Your Travel Plan"
    )

    # --------------------------------------------------------
    # BUTTON ACTION
    # --------------------------------------------------------

    generate_button.click(
        fn=generate_travel_plan,
        inputs=[
            destination,
            days,
            budget,
            travelers,
            interests,
            food,
            travel_style
        ],
        outputs=output
    )


# ============================================================
# STEP 6: LAUNCH APPLICATION
# ============================================================

app.launch(share=True)
```

```python
# ==========================================
# ✈️ AI TRAVEL PLANNER USING TRANSFORMER
# Simple Google Colab Version
# ==========================================

!pip -q install tensorflow gradio

import tensorflow as tf
import gradio as gr

# -----------------------------
# 1. Small Travel Dataset
# -----------------------------

travel_data = {
    "goa": [
        "Day 1: Visit Baga Beach and enjoy local food.",
        "Day 2: Visit Fort Aguada and Calangute Beach.",
        "Day 3: Visit Dudhsagar Falls and enjoy shopping."
    ],

    "dubai": [
        "Day 1: Visit Burj Khalifa and Dubai Mall.",
        "Day 2: Enjoy Desert Safari and Arabian food.",
        "Day 3: Visit Jumeirah Beach and Dubai Marina."
    ],

    "manali": [
        "Day 1: Visit Mall Road and Hadimba Temple.",
        "Day 2: Visit Solang Valley and enjoy adventure activities.",
        "Day 3: Visit Rohtang Pass and enjoy mountain views."
    ],

    "paris": [
        "Day 1: Visit Eiffel Tower and Seine River.",
        "Day 2: Visit Louvre Museum and Arc de Triomphe.",
        "Day 3: Explore local markets and enjoy French food."
    ]
}

# -----------------------------
# 2. Simple Transformer Layer
# -----------------------------

class SimpleTransformer(tf.keras.layers.Layer):

    def __init__(self):
        super().__init__()

        self.attention = tf.keras.layers.MultiHeadAttention(
            num_heads=2,
            key_dim=16
        )

        self.normalization = tf.keras.layers.LayerNormalization()

        self.dense = tf.keras.layers.Dense(32, activation="relu")

    def call(self, inputs):

        attention_output = self.attention(
            inputs,
            inputs
        )

        output = self.normalization(
            inputs + attention_output
        )

        return self.dense(output)


# Create Transformer
transformer = SimpleTransformer()

# -----------------------------
# 3. Travel Planner
# -----------------------------

def generate_plan(destination, days, budget, interests):

    destination = destination.lower().strip()

    if destination not in travel_data:
        return """
❌ Destination not available.

Try:
• Goa
• Dubai
• Manali
• Paris
"""

    days = int(days)

    # Create simple input for Transformer
    x = tf.random.normal((1, days, 16))

    # Process through Transformer
    transformer(x)

    places = travel_data[destination]

    result = f"""
# ✈️ AI Travel Plan

### 📍 Destination: {destination.title()}
### 📅 Duration: {days} days
### 💰 Budget: {budget}
### ❤️ Interests: {interests}

---

"""

    for i in range(days):

        plan = places[i % len(places)]

        result += f"## 📅 {plan}\n\n"

    result += """
---

### 🎒 Travel Tips

• Carry important documents
• Carry comfortable clothes
• Keep emergency money
• Check the weather before travelling
• Carry a power bank

### ✈️ Have a safe journey!
"""

    return result


# -----------------------------
# 4. Gradio Interface
# -----------------------------

app = gr.Interface(

    fn=generate_plan,

    inputs=[

        gr.Textbox(
            label="📍 Destination",
            placeholder="Example: Goa"
        ),

        gr.Number(
            label="📅 Number of Days",
            value=3
        ),

        gr.Textbox(
            label="💰 Budget",
            placeholder="Example: ₹30,000"
        ),

        gr.Textbox(
            label="❤️ Interests",
            placeholder="Example: beaches, shopping, adventure"
        )
    ],

    outputs=gr.Markdown(),

    title="✈️ AI Travel Planner",

    description=
    "A simple Transformer-based travel planning system."
)

app.launch(share=True)
```
